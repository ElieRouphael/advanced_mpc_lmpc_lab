## B. Engineering questions

- **Solver failure:** what should a real controller do when the online optimization is infeasible or times out?
- **Model mismatch:** which guarantees survive if $x^+=Ax+Bu$ is only approximate?
- **Constraint softening:** which constraints can be softened and which are safety-critical hard constraints?
- **State estimation:** how would a Kalman filter enter the architecture when the full state is not measured?
- **Delay:** how would actuation/communication delay change the prediction model?
- **Uncertainty:** when would you choose robust/tube MPC, stochastic MPC, or chance constraints?
- **Learning:** how do you prevent a learned local model from extrapolating outside its data support?
